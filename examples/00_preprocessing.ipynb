# AstroPrism Tutorial: Preprocessing
Before loading data with AstroPrism, each FITS file needs a **PSF** extension appended.
This notebook uses `append_psf_extension` to embed PSF arrays from `.npz` files into the tutorial JWST MIRI FITS files.

## Imports

```python
from pathlib import Path

from astropy.io import fits

from astroprism.io import append_psf_extension
```

## Paths
Point to the tutorial FITS data and the matching PSF `.npz` files.

```python
repo_root = Path("..")
hack_sample = repo_root / "data" / "tutorial" / "hack_sample"

data_dir = hack_sample / "data"
psf_dir = hack_sample / "psfs"

bands = ["f770w", "f1000w", "f1130w", "f2100w"]

for band in bands:
    data_file = data_dir / f"hlsp_phangs-jwst_jwst_miri_ngc1566_{band}_v1p0p1_img.fits"
    psf_file = psf_dir / f"{band.upper()}.npz"
    print(f"{band}:  data={data_file.name}  psf={psf_file.name}")
```

## Embed PSFs
Loop over the four MIRI bands and append the PSF extension to each FITS file.

```python
output_dir = hack_sample / "data_full"

for band in bands:
    data_file = str(data_dir / f"hlsp_phangs-jwst_jwst_miri_ngc1566_{band}_v1p0p1_img.fits")
    psf_file = str(psf_dir / f"{band.upper()}.npz")
    output_file = str(output_dir / f"hlsp_phangs-jwst_jwst_miri_ngc1566_{band}_v1p0p1_img.fits")

    hdul = append_psf_extension(data_file, psf_file, output_file, overwrite=True)
    print(f"{band}: embedded PSF — extensions: {[hdu.name for hdu in hdul]}")
    hdul.close()
```

## Verify
Re-open one of the files and confirm the PSF extension is present.

```python
with fits.open(str(output_dir / "hlsp_phangs-jwst_jwst_miri_ngc1566_f770w_v1p0p1_img.fits")) as hdul:
    hdul.info()
    print(f"\nPSF shape: {hdul['PSF'].data.shape}")
```
